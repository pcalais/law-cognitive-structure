````python
# Caminho para o arquivo
caminho_prompt = "prompts/generate_irac_prompt.txt"

# Lê o conteúdo do arquivo como string
with open(caminho_prompt, "r", encoding="utf-8") as arquivo:
    prompt_text = arquivo.read()

# Exibe parte do conteúdo
print(prompt_text[:500])  # Mostra os primeiros 500 caracteres
````

````python
import pandas as pd
import time
from tqdm import tqdm
import os
import json

def create_irac(oab_questions_df, model_name, N, output_path):
    total_questions = len(oab_questions_df)

    # Carrega resultados anteriores, se existirem
    if os.path.exists(output_path):
        existing_df = pd.read_csv(output_path)
        processed_keys = set(zip(existing_df['pdf_filename'], existing_df['questao']))
    else:
        existing_df = pd.DataFrame()
        processed_keys = set()

    # Cria chave única e filtra registros ainda não processados
    oab_questions_df['unique_key'] = list(zip(oab_questions_df['pdf_filename'], oab_questions_df['questao']))
    remaining_df = oab_questions_df[~oab_questions_df['unique_key'].isin(processed_keys)]

    # Estatísticas de progresso
    total_processed = len(processed_keys)
    total_remaining = len(remaining_df)
    pct_processed = (total_processed / total_questions) * 100
    pct_remaining = (total_remaining / total_questions) * 100

    print(f"Total de questões: {total_questions}")
    print(f"Já processadas: {total_processed} ({pct_processed:.1f}%)")
    print(f"Faltando: {total_remaining} ({pct_remaining:.1f}%)")

    # Se não há mais registros, retorna o DataFrame completo
    if remaining_df.empty:
        print("Todas as questões já foram processadas.")
        return existing_df

    # Seleciona N registros ainda não processados
    amostra_df = remaining_df.sample(n=min(N, total_remaining)).reset_index(drop=True)

    # Inicializa modelo
    genai.configure(api_key=API_KEY)
    model = genai.GenerativeModel(model_name)

    new_answers_df = pd.DataFrame()

    for index, row in tqdm(amostra_df.iterrows(), total=len(amostra_df), desc=f"Executando questões {model_name}"):
        question_body = f'''
        {row['enunciado']}

        A) {row['A']}.Gabarito comentado: {row['justificativa_A']}
        B) {row['B']} Gabarito comentado: {row['justificativa_B']}
        C) {row['C']} Gabarito comentado: {row['justificativa_C']}
        D) {row['D']} Gabarito comentado: {row['justificativa_D']}
        '''

        response = model.generate_content(prompt_text + "\n" + question_body)
        resposta_texto = response.text


        # Tenta fazer parsing do JSON da resposta
        try:
            irac = json.loads(resposta_texto.replace("```json", "").replace("```", ""))
            issue = irac.get("Issue", "")
            rule = irac.get("Rule", "")
            application = irac.get("Application", "")
            conclusion = irac.get("Conclusion", "")
        except Exception as e:
            print(f"[Erro no parsing JSON] questão {row['questao']}: {e}", resposta_texto)
            issue = rule = application = conclusion = ""

        new_answers_df = pd.concat([
            new_answers_df,
            pd.DataFrame([{
                "question_id": row['question_id'],
                "pdf_filename": row['pdf_filename'],
                "materia": row['materia'],
                "prova": row['prova'],
                "questao": row['questao'],
                "enunciado": row['enunciado'],
                "A": row['A'],
                "justificativa_A": row['justificativa_A'],
                "B": row['B'],
                "justificativa_B": row['justificativa_B'],
                "C": row['C'],
                "justificativa_C": row['justificativa_C'],
                "D": row['D'],
                "justificativa_D": row['justificativa_D'],
                "Issue": issue,
                "Rule": json.dumps(rule),
                "Application": application,
                "Conclusion": conclusion,
                "correct": row['correct']
            }])
        ], ignore_index=True)

        time.sleep(4)

    # Salva resultados atualizados
    updated_df = pd.concat([existing_df, new_answers_df], ignore_index=True)
    updated_df.to_csv(output_path, index=False, encoding='utf-8-sig')

    return updated_df
````

````python
# Função recursiva para coletar todas as ocorrências da chave "lei"
def encontrar_todas_leis(json_string):
    dados = json.loads(json_string)
    leis_encontradas = []

    for categoria in dados.values():
        for item in categoria:
            if isinstance(item, dict) and "lei" in item:
                leis_encontradas.append(item["lei"])

    return leis_encontradas


rule = """
{"constituicao_federal": [{"lei": "lei tal da CF"}], "codigo_penal": [], "codigo_processo_penal": [], "clt": [], "codigo_civil": [{"lei": "Art. 733 do C\u00f3digo Civil", "conteudo": "Nos contratos de transporte cumulativo, cada transportador se obriga a cumprir o contrato relativamente ao respectivo percurso, respondendo pelos danos nele causados a pessoas e coisas."}], "codigo_processo_civil": [], "sumula_stj": [], "estatuto_oab": [], "codigo_tributario_nacional": [], "principio_geral_do_direito": [], "outros": [{"lei": "Art. 2\u00ba da Lei n\u00ba 9.611/98", "conteudo": "Transporte Multimodal de Cargas \u00e9 aquele que, regido por um \u00fanico contrato, utiliza duas ou mais modalidades de transporte, desde a origem at\u00e9 o destino."}, {"lei": "Conceito de Contrato de Transporte Combinado", "conteudo": "No contrato de transporte combinado o transportador assume a obriga\u00e7\u00e3o de transportar a coisa no seu trecho, assumindo, perante o cliente, a obriga\u00e7\u00e3o de contratar terceiros para a continua\u00e7\u00e3o do transporte em outros trechos."}, {"lei": "Conceito de Transporte de Fato", "conteudo": "O transporte de fato se d\u00e1 quando realizado por pessoa distinta do transportador constante do contrato, autorizado por este a efetuar o transporte integral ou parcialmente."}]}
"""
assert len(encontrar_todas_leis(rule)) == 5
````

````python
import pandas as pd
import json
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Converter a coluna 'Rule' (string JSON) para dicionário Python
oab_with_irac_df["Rule_dict"] = oab_with_irac_df["Rule"].apply(json.loads)

# 2. Contar o número de leis por fonte do direito (nível 1 do JSON)
def contar_leis_por_fonte(rule_dict):
    return {fonte: len(itens) for fonte, itens in rule_dict.items() if isinstance(itens, list)}

oab_with_irac_df["leis_por_fonte"] = oab_with_irac_df["Rule_dict"].apply(contar_leis_por_fonte)

# 3. Expandir a coluna 'leis_por_fonte' em colunas separadas
leis_df = oab_with_irac_df[["materia", "leis_por_fonte"]].copy()
leis_expandidas = leis_df["leis_por_fonte"].apply(pd.Series)
leis_expandidas["materia"] = leis_df["materia"]

# 4. Agrupar por matéria e calcular a média de leis por fonte
media_leis_por_materia = leis_expandidas.groupby("materia").mean(numeric_only=True)

# 5. Calcular o total médio de leis por matéria e ordenar
media_leis_por_materia["total_medio"] = media_leis_por_materia.sum(axis=1)
media_leis_por_materia = media_leis_por_materia.sort_values("total_medio", ascending=False)
media_leis_ordenadas = media_leis_por_materia.drop(columns="total_medio")

# 6. Plotar o heatmap ordenado
plt.figure(figsize=(12, 6))
sns.heatmap(media_leis_ordenadas, annot=True, cmap="YlGnBu", fmt=".2f")
plt.title("Número médio de leis por fonte do direito e por matéria (ordenado)")
plt.xlabel("Fonte primária do direito (1º nível do JSON)")
plt.ylabel("Matéria")
plt.tight_layout()
plt.show()
````

````python
import pandas as pd
import json
import random

# Carrega o DataFrame
oab_with_irac_df = pd.read_csv("../data/processed/oab_with_irac.csv")

dataframes_por_linha = []
num_linhas = len(oab_with_irac_df)
indices = list(range(num_linhas))

for idx, row in oab_with_irac_df.iterrows():
    nova_linha = row.copy()

    # Extrai "lei" da linha atual para a coluna A
    rule_data = row['Rule']
    leis_A = encontrar_todas_leis(rule_data)
    nova_linha['A'] = '; '.join(str(lei) for lei in leis_A) if leis_A else ''

    # Filtra índices com a mesma matéria (exceto o atual)
    materia_atual = row['materia']
    mesma_materia_indices = oab_with_irac_df[
        (oab_with_irac_df['materia'] == materia_atual) & (oab_with_irac_df.index != idx)
    ].index.tolist()


    # Se houver ao menos 3, usa eles; senão, usa quaisquer outros
    if len(mesma_materia_indices) >= 3:
        aleatorios = random.sample(mesma_materia_indices, 3)
    else:
        possiveis_indices = [i for i in indices if i != idx]
        aleatorios = random.sample(possiveis_indices, 3)

    # Preenche colunas B, C e D
    for col, i_aleatorio in zip(['B', 'C', 'D'], aleatorios):
        rule_aleatorio = oab_with_irac_df.loc[i_aleatorio, 'Rule']
        leis_aleatorias = encontrar_todas_leis(rule_aleatorio)
        valor_concat = '; '.join(str(lei) for lei in leis_aleatorias) if leis_aleatorias else ''
        nova_linha[col] = valor_concat

    df_novo = pd.DataFrame([nova_linha])
    dataframes_por_linha.append(df_novo)

# Junta tudo e salva
df_testa_lei = pd.concat(dataframes_por_linha, ignore_index=True)
df_testa_lei['correct'] = 'A'
df_testa_lei.to_csv("../data/processed/oab_guess_the_rule.csv", index=False, encoding='utf-8-sig')
````
