```python
import pandas as pd
import json
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 1. Converte a coluna Rule para dicionário (se ainda não estiver)
oab_exam_df['Rule_dict'] = oab_exam_df['Rule'].apply(
    lambda x: json.loads(x) if isinstance(x, str) else x
)

# 2. Extrai automaticamente todas as chaves (tipos de lei) encontradas nos JSONs
all_keys = sorted({
    key
    for d in oab_exam_df['Rule_dict']
    for key in d.keys()
})

# 3. Cria colunas com a contagem de itens em cada tipo de lei
for law in all_keys:
    oab_exam_df[law] = oab_exam_df['Rule_dict'].apply(lambda x: len(x.get(law, [])))

# 4. Calcula a média de leis por tipo, agrupando por 'materia'
avg_laws = oab_exam_df.groupby('materia')[all_keys].mean().round(2)

# 5. Cria uma máscara onde o valor é 0
mask = avg_laws == 0

# 6. Plota o heatmap com as células 0 ocultas (em branco)
plt.figure(figsize=(12, 6))
sns.heatmap(
    avg_laws,
    annot=True,
    cmap='Blues',
    fmt=".2f",
    cbar=True,
    linewidths=0.5,
    linecolor='white',
    mask=mask
)
plt.title('Média de Leis Citadas por Tipo de Lei e Matéria (OAB)')
plt.xlabel('Tipo de Lei')
plt.ylabel('Matéria')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import json
import matplotlib.pyplot as plt
from collections import Counter

# 1. Converte a coluna Rule para dicionário (se ainda não estiver)
oab_exam_df['Rule_dict'] = oab_exam_df['Rule'].apply(
    lambda x: json.loads(x) if isinstance(x, str) else x
)

# 2. Extrai todos os valores do campo "numero" dentro de qualquer tipo de lei
all_laws = []
for rules in oab_exam_df['Rule_dict']:
    for law_list in rules.values():
        for item in law_list:
            numero = item.get('numero')
            if numero:
                all_laws.append(numero.strip())

# 3. Conta as leis mais frequentes
law_counts = Counter(all_laws)
top10 = law_counts.most_common(50)

# 4. Prepara para plotagem
labels, values = zip(*top10)

plt.figure(figsize=(10, 6))
plt.barh(labels, values, color='steelblue')
plt.xlabel('Número de Ocorrências')
plt.ylabel('Lei / Artigo')
plt.title('Top 10 Leis mais Citadas no Campo "Rule"')
plt.gca().invert_yaxis()  # coloca a mais citada no topo
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calcula a média de rule_count por matéria e ordena do menor para o maior
order = (
    oab_exam_df.groupby('materia')['rule_count']
    .mean()
    .sort_values()
    .index
)

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=oab_exam_df,
    x='materia',
    y='rule_count',
    order=order,          # aplica a ordenação
    palette='Blues',
    showmeans=True,
    meanprops={"marker":"o", "markerfacecolor":"white", "markeredgecolor":"black"}
)

plt.xlabel('Matéria')
plt.ylabel('Número de Regras (rule_count)')
plt.title('Distribuição de rule_count por Matéria (ordenado pela média)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from nltk.corpus import stopwords
import nltk
import json
import re

# baixar stopwords (somente na primeira vez)
nltk.download("stopwords")

stopwords_pt = set(stopwords.words("portuguese"))

cols = ["Facts", "Issue", "Rule", "Application", "Conclusion"]

def extract_json_values(text):
    """Se for JSON válido, retorna apenas os valores. Senão, retorna o próprio texto."""
    try:
        obj = json.loads(text)
        # pega só os valores do JSON
        return " ".join(str(v) for v in obj.values())
    except:
        return text  # não era JSON, retorna o texto normal

for col in cols:
    # Tratamento especial para Rule
    if col == "Rule":
        processed = oab_exam_df[col].dropna().astype(str).apply(extract_json_values)
        text = " ".join(processed.values)
    else:
        text = " ".join(oab_exam_df[col].dropna().astype(str).values)

    wc = WordCloud(
        width=800,
        height=400,
        background_color="white",
        stopwords=stopwords_pt,
        collocations=True
    ).generate(text)

    plt.figure(figsize=(10,5))
    plt.title(f"WordCloud - {col}", fontsize=16)
    plt.imshow(wc, interpolation="bilinear")
    plt.axis("off")
    plt.show()
```
